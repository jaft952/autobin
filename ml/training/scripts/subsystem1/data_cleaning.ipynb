```python
import os
import glob

def remove_specific_dataset_files(dataset_dir, targets_list):
    print("Starting targeted file removal...")
    
    # Track metrics for the final summary
    removed_images = 0
    removed_labels = 0
    
    # Standard image extensions to scan for
    valid_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".PNG", ".JPG"]

    for target_name in targets_list:
        # Strip any trailing extension if you accidentally pasted it with one
        base_target = os.path.splitext(target_name)[0]
        
        # 1. Search and destroy matching image files across all splits
        # Looks inside train/images, val/images, test/images
        img_pattern = os.path.join(dataset_dir, "**", "images", f"{base_target}.*")
        found_images = glob.glob(img_pattern, recursive=True)
        
        for img_path in found_images:
            # Extra safety check to ensure we only delete actual image formats
            if os.path.splitext(img_path)[1] in valid_extensions:
                try:
                    os.remove(img_path)
                    print(f"Deleted Image: {os.path.basename(img_path)}")
                    removed_images += 1
                except Exception as e:
                    print(f"Error deleting image {img_path}: {e}")

        # 2. Search and destroy matching label text files
        # Looks inside train/labels, val/labels, test/labels
        lbl_pattern = os.path.join(dataset_dir, "**", "labels", f"{base_target}.txt")
        found_labels = glob.glob(lbl_pattern, recursive=True)
        
        for lbl_path in found_labels:
            try:
                os.remove(lbl_path)
                print(f"Deleted Label: {os.path.basename(lbl_path)}")
                removed_labels += 1
            except Exception as e:
                print(f"Error deleting label {lbl_path}: {e}")

    print("\n==============================================")
    print("CLEANUP SUMMARY")
    print("==============================================")
    print(f"Total Image Files Erased: {removed_images}")
    print(f"Total Label Files Erased: {removed_labels}")
    print("==============================================")

bad_files = [
    "pair_50_ds2_IMG_6879_MOV-0008_jpg.rf.IQbf4jwtE609Xwnca2Ki",
    "pair_374_ds1_IMG_7574_jpg.rf.McNS0f92cWsXWJRaPzu0_aug_v1",
    "pair_415_ds2_IMG_6877_MOV-0011_jpg.rf.XnL9W0LQZkktSh6EoObD_aug_v1",
    "pair_424_ds2_IMG_6878_MOV-0026_jpg.rf.rFyDQBwuOuOJxBYNlxYN_aug_v1",
    "pair_424_ds2_IMG_6878_MOV-0026_jpg.rf.rFyDQBwuOuOJxBYNlxYN_aug_v2",
    "pair_424_ds2_IMG_6878_MOV-0026_jpg.rf.rFyDQBwuOuOJxBYNlxYN",
    "pair_431_ds2_IMG_6879_MOV-0002_jpg.rf.nVf0kQ39chrksquJTKHb",
    "pair_439_ds2_IMG_6880_MOV-0002_jpg.rf.iHXWv07DCkyIUjDqfY3I",
    "pair_443_ds2_IMG_6881_MOV-0001_jpg.rf.bLgEYvc0FutlDUyh5Kvv"
]

remove_specific_dataset_files(
    dataset_dir="../data/tin.yolov8", 
    targets_list=bad_files
)
```
