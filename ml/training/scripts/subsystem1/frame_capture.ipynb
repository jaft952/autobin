```python
import cv2
import os
from datetime import datetime

# ── CONFIG ──────────────────────────────────────────────────────────────────
CAMERA_INDEX = 0
SAVE_PATH    = r"../data/captures"
# ────────────────────────────────────────────────────────────────────────────

os.makedirs(SAVE_PATH, exist_ok=True)
print("Press  S  to snap  |  Q  to quit")

cap = cv2.VideoCapture(CAMERA_INDEX)
snap_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    preview = frame.copy()
    cv2.putText(preview, f"S=snap  Q=quit  saved={snap_count}",
                (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)
    cv2.imshow("Snap", preview)
    key = cv2.waitKey(1) & 0xFF

    if key == ord("q"):
        break
    if key == ord("s"):
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
        filename  = os.path.join(SAVE_PATH, f"frame_{timestamp}.jpg")
        cv2.imwrite(filename, frame)
        snap_count += 1
        print(f"[{snap_count}] Saved -> {filename}")

cap.release()
cv2.destroyAllWindows()
print(f"Done. {snap_count} frame(s) saved.")
```
