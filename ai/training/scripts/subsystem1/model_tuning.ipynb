```python

import os
import torch
from ultralytics import YOLO

print(f"CUDA Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Using GPU Device: {torch.cuda.get_device_name(0)}")
else:
    print("WARNING: CUDA GPU not found. Training will run slowly on the CPU.")
    

torch.backends.cudnn.enabled = True
torch.backends.cudnn.benchmark = False
torch.backends.cudnn.deterministic = True
```

```python
model = YOLO("../runs/model_tuning_24062026/weights/best.pt")

model.train(
    data="../data/tin.yolov8.5/data.yaml",
    epochs=50,
    imgsz=640,
    device="0" if torch.cuda.is_available() else "cpu",
    batch=16,
    patience=20,
    project="../",
    name="model_tuning"
)
```
