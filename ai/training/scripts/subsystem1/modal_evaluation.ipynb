```python
from pathlib import Path

# === RUN THIS FIRST: repo-root anchor so paths resolve from any working dir ===
# Walks up until it finds the repo root (the folder containing both ai/ and src/).
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "ai").is_dir() and (p / "src").is_dir()), Path.cwd())

DATASETS = ROOT / "ai" / "data" / "datasets"                      # datasets root
RUNS     = ROOT / "ai" / "runs"                                   # training/eval output
DEPLOY   = ROOT / "ai" / "models" / "subsystem1" / "production"   # deployed .pt models

print("ROOT     =", ROOT)
print("DATASETS =", DATASETS)
print("RUNS     =", RUNS)
print("DEPLOY   =", DEPLOY)
```

```python
from ultralytics import YOLO

model = YOLO(str(RUNS / "model_training" / "weights" / "best.pt"))
metrics = model.val(split="test", project=str(RUNS), name="model_evaluation")

print(f"Final Test Accuracy (mAP50): {metrics.box.map50 * 100:.2f}%")
```

```python
import shutil
import os
from datetime import datetime

source_path = str(RUNS / "model_training" / "weights" / "best.pt")
dest_dir = str(DEPLOY)
today = datetime.now().strftime("%d%m%Y")
dest_path = os.path.join(dest_dir, f"inference_{today}.pt")

os.makedirs(dest_dir, exist_ok=True)

shutil.copy(source_path, dest_path)
print(f"Successfully deployed champion model to: {dest_path}")
```

```python
from datetime import datetime

model_dir = str(RUNS / "model_training")
evaluation_dir = str(RUNS / "model_evaluation")

if os.path.exists(model_dir):
    today = datetime.now().strftime("%d%m%Y")
    new_model_dir = str(RUNS / f"model_training_{today}")
    os.rename(model_dir, new_model_dir)
    print(f"Renamed previous model folder to: {new_model_dir}")

if os.path.exists(evaluation_dir):
    today = datetime.now().strftime("%d%m%Y")
    new_evaluation_dir = str(RUNS / f"model_evaluation_{today}")
    os.rename(evaluation_dir, new_evaluation_dir)
    print(f"Renamed previous evaluation folder to: {new_evaluation_dir}")
```
