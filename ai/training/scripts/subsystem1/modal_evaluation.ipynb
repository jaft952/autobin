```python
from ultralytics import YOLO

model = YOLO("../runs/model_training/weights/best.pt")
metrics = model.val(split="test", project="../", name="model_evaluation")

print(f"Final Test Accuracy (mAP50): {metrics.box.map50 * 100:.2f}%")
```

```python
import shutil
import os
from datetime import datetime

source_path = "../runs/model_training/weights/best.pt"
dest_dir = "../src/models"
today = datetime.now().strftime("%d%m%Y")
dest_path = os.path.join(dest_dir, f"inference_{today}.pt")

os.makedirs(dest_dir, exist_ok=True)

shutil.copy(source_path, dest_path)
print(f"Successfully deployed champion model to: {dest_path}")
```

```python
from datetime import datetime

model_dir = "../runs/model_training"
evaluation_dir = "../runs/model_evaluation"

if os.path.exists(model_dir):
    today = datetime.now().strftime("%d%m%Y")
    new_model_dir = f"../runs/model_training_{today}"
    os.rename(model_dir, new_model_dir)
    print(f"Renamed previous model folder to: {new_model_dir}")

if os.path.exists(evaluation_dir):
    today = datetime.now().strftime("%d%m%Y")
    new_evaluation_dir = f"../runs/model_evaluation_{today}"
    os.rename(evaluation_dir, new_evaluation_dir)
    print(f"Renamed previous evaluation folder to: {new_evaluation_dir}")
```
