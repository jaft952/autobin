```python
import os
from PIL import Image
from pillow_heif import register_heif_opener

register_heif_opener()

def convert_heic_folder_to_output(input_folder, output_folder, quality=100):
    """Converts HEIC files from an input folder and saves JPEGs to an output folder."""
    if not os.path.exists(input_folder):
        return

    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        print(f"Created output directory: {output_folder}")

    converted_count = 0
    
    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".heic"):
            heic_path = os.path.join(input_folder, filename)
            
            jpg_filename = os.path.splitext(filename)[0] + ".jpg"
            jpg_path = os.path.join(output_folder, jpg_filename)
            
            try:
                image = Image.open(heic_path)
                rgb_image = image.convert("RGB")
                rgb_image.save(jpg_path, "JPEG", quality=quality)
                
                print(f"Converted: {filename} -> {os.path.join(output_folder, jpg_filename)}")
                converted_count += 1
                
            except Exception as e:
                print(f"Failed to convert {filename}. Error: {e}")
                
    print(f"\nTask finished. Total files converted and moved: {converted_count}")
    
source_folder = "../data/img_heic"
destination_folder = "../data/img_jpg"

convert_heic_folder_to_output(source_folder, destination_folder)
```
