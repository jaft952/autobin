```python
from pathlib import Path

# === RUN THIS FIRST: repo-root anchor so paths resolve from any working dir ===
# Walks up until it finds the repo root (the folder containing both ai/ and src/).
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "ai").is_dir() and (p / "src").is_dir()), Path.cwd())

DATASETS = ROOT / "ai" / "data" / "datasets"                      # datasets root
RUNS     = ROOT / "ai" / "runs"                                   # training/eval output
DEPLOY   = ROOT / "ai" / "models" / "subsystem1" / "production"   # deployed .pt models

print("ROOT     =", ROOT)
print("DATASETS =", DATASETS)
print("RUNS     =", RUNS)
print("DEPLOY   =", DEPLOY)
```

```python
import os
import torch
from ultralytics import YOLO

print(f"CUDA Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Using GPU Device: {torch.cuda.get_device_name(0)}")
else:
    print("WARNING: CUDA GPU not found. Training will run slowly on the CPU.")
```

```python
model = YOLO("yolov8n.yaml") 

model.load("yolov8n.pt") 

model.train(
    data=str(DATASETS / "tin.yolov8" / "data.yaml"),
    epochs=50,                          
    imgsz=640,
    device="0" if torch.cuda.is_available() else "cpu",
    project=str(RUNS),
    name="model_training"
)
```

```python
from IPython.display import Image, display

results_path = str(RUNS / "model_training" / "results.png")

if os.path.exists(results_path):
    display(Image(filename=results_path))
else:
    print(f"Error: Could not find '{results_path}'")
```
