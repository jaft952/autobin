```python
import os
import torch
from ultralytics import YOLO

print(f"CUDA Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Using GPU Device: {torch.cuda.get_device_name(0)}")
else:
    print("WARNING: CUDA GPU not found. Training will run slowly on the CPU.")
```

```python
model = YOLO("yolov8n.yaml") 

model.load("../yolov8n.pt") 

model.train(
    data="../data/tin.yolov8/data.yaml",
    epochs=50,                          
    imgsz=640,
    device="0" if torch.cuda.is_available() else "cpu",
    project="../",
    name="model_training"
)
```

```python
from IPython.display import Image, display

results_path = "../runs/model_training/results.png"

if os.path.exists(results_path):
    display(Image(filename=results_path))
else:
    print(f"Error: Could not find '{results_path}'")
```
