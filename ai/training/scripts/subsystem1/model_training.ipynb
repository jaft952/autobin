```python
import gc
import time
import torch
from ultralytics import YOLO

if torch.cuda.is_available():
    print(f"Device : {torch.cuda.get_device_name(0)}")
    print(f"VRAM   : {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB\n")
else:
    print("WARNING: CUDA GPU not found. Training will run slowly on the CPU.")

torch.backends.cudnn.enabled = True
torch.backends.cudnn.benchmark = False
torch.backends.cudnn.deterministic = False

DATA_YAML   = r"../../../data/datasets/augmented_tin/data.yaml"
PROJECT_DIR = r"../../../../"
EPOCHS      = 200
IMGSZ       = 640
BATCH       = -1
PATIENCE    = 20
DEVICE      = 0 if torch.cuda.is_available() else "cpu"
SEED        = 42

def clear_vram():
    global model, results
    if 'model' in globals(): del model # type: ignore
    if 'results' in globals(): del results # type: ignore
    gc.collect()
    torch.cuda.empty_cache()
    time.sleep(2)

clear_vram()
```

```python
start = time.time()
model = YOLO("yolo11n-seg.pt")

results = model.train(
    data     = DATA_YAML,
    epochs   = EPOCHS,
    imgsz    = IMGSZ,
    batch    = BATCH,
    patience = PATIENCE,
    device   = DEVICE,
    project  = PROJECT_DIR,
    name     = "yolo11n-seg",
    seed     = SEED,
)

elapsed = time.time() - start
h, m = int(elapsed // 3600), int((elapsed % 3600) // 60)
print(f"\nyolo11n-seg done in {h}h {m}m")
torch.cuda.empty_cache()
```
